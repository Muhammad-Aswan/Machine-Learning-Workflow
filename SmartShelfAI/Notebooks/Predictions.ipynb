```python
new_data = pd.DataFrame({
    "Category": ["Beverages"],
    "Supplier": ["Supplier_A"],
    "Region": ["North"],
    "Store_Size": ["Large"],
    "Snapshot_Date": ["2026-07-29"],
    "Season": ["Summer"],
    "Current_Stock": [120],
    "Daily_Sales": [35],
    "Reorder_Level": [50],
    "Lead_Time_Days": [7],
    "Unit_Price": [4.99],
    "Discount_Percent": [10],
    "Shelf_Capacity": [200],
    "Promotion": [1],
    "Holiday_Week": [0],
    "Days_Since_Last_Restock": [5],
    "Customer_Demand_Index": [0.82],
    "Weather_Impact": ["Low"],
    "Is_Perishable": [1],
    "Stock_Coverage_Days": [3.4],
    "Price_per_Demand": [6.08],
    "Supplier_Lead_Reliability": [0.95],
    "Store_Age_Years": [8],
    "Last_Audit_Score": [91],
    "Competitor_Price_Index": [1.03],
    "Random_Noise_A": [0.12]
})
```

```python
prediction = rf.predict(new_data)[0]
probability = rf.predict_proba(new_data)[0]

print(f"Prediction: {prediction}")

if prediction == 1:
    print("Stockout Risk: YES")
else:
    print("Stockout Risk: NO")

print(f"Probability of No Stockout : {probability[0]:.2%}")
print(f"Probability of Stockout    : {probability[1]:.2%}")
```
